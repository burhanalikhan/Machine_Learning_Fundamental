```python
from sklearn.datasets import fetch_openml
mnist = fetch_openml('mnist_784', version=1)
mnist.keys()
```

```python
X, y = mnist["data"], mnist["target"]
```

```python
import matplotlib as mpl
import matplotlib.pyplot as plt

some_digit = X.iloc[0].values  # Get the first row and convert to NumPy array
some_digit_image = some_digit.reshape(28, 28)

plt.imshow(some_digit_image, cmap=mpl.cm.binary, interpolation="nearest")
plt.axis("off")
plt.show()
```

```python
import numpy as np
y = y.astype(np.uint8)
```

```python
 X_train, X_test, y_train, y_test = X[:60000], X[60000:], y[:60000], y[60000:]
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn_clf = KNeighborsClassifier()
knn_clf.fit(X_train, y_train)
```

```python
knn_clf.predict([some_digit])
```

```python
from sklearn.model_selection import GridSearchCV
param_grid = [{'weights': ["uniform", "distance"], 'n_neighbors': [3, 4, 5]}]
knn_clf2 = KNeighborsClassifier()
grid_search = GridSearchCV(knn_clf2, param_grid, cv=5, verbose=3)
grid_search.fit(X_train, y_train)
```

```python
grid_search.best_params_
```

```python
from sklearn.metrics import accuracy_score

y_pred = grid_search.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
grid_search.best_score_
```

```python
from scipy.ndimage.interpolation import shift
```

```python
def shift_image(image, dx, dy):
    image = image.reshape((28, 28))
    shifted_image = shift(image, [dy, dx], cval=0, mode="constant")
    return shifted_image.reshape([-1])
```

```python
image_subset = X_train.iloc[:1000].values  # Convert first 1000 rows to NumPy array
shifted_images_down = np.array([shift_image(img, 0, 5) for img in image_subset])
shifted_images_left = np.array([shift_image(img, -5, 0) for img in image_subset])
```

```python
X_train = np.array(X_train)  # Ensure it's a NumPy array
y_train = np.array(y_train)  # Ensure labels are also a NumPy array
X_train_augmented = [image for image in X_train]
y_train_augmented = [label for label in y_train]

for dx, dy in ((1, 0), (-1, 0), (0, 1), (0, -1)):
    for image, label in zip(X_train, y_train):
        X_train_augmented.append(shift_image(image, dx, dy))
        y_train_augmented.append(label)

X_train_augmented = np.array(X_train_augmented)
y_train_augmented = np.array(y_train_augmented)
```

```python
shuffle_idx = np.random.permutation(len(X_train_augmented))
X_train_augmented = X_train_augmented[shuffle_idx]
y_train_augmented = y_train_augmented[shuffle_idx]
```

```python
knn_clf = KNeighborsClassifier(**grid_search.best_params_)
knn_clf.fit(X_train_augmented, y_train_augmented)
```

```python
y_pred = knn_clf.predict(X_test)
accuracy_score(y_test, y_pred)
```
